```python
#This code was generated using ChatGPT solely for the purposes of education.
import sympy as sp

def compute_matrix_power():
    """
    Interactive tool to compute A^k * v for a 4x4 matrix A and a 1x4 column vector v using sympy.
    """
    # Matrix A
    matrixA = [
        [0, 0, 1, 1/2],
        [1/3, 0, 0, 0],
        [1/3, 1/2, 0, 1/2],
        [1/3, 1/2, 0, 0]
    ]
    A = sp.Matrix(matrixA)
    
    # Vector v
    vectorv = [
    [1/4],
    [1/4],
    [1/4],
    [1/4]
]
    v = sp.Matrix(vectorv)

    
    while True:
        try:
            k = int(input("\nEnter the power k (non-negative integer): "))
            if k < 0:
                print("Please enter a non-negative integer for k.")
                continue
        except ValueError:
            print("Invalid input. Please enter an integer for k.")
            continue
        
        # Compute A^k
        Ak = A**k
        
        # Compute A^k * v
        result = Ak * v
        decimal_result = result.evalf()  # Convert to decimals
        
        print(f"\nA^{k} * v (decimal values) =")
        sp.pprint(decimal_result)
        
        cont = input("\nDo you want to compute for another k? (yes/no): ").strip().lower()
        if cont != 'yes':
            break

if __name__ == "__main__":
    compute_matrix_power()

```
